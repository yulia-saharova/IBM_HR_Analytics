# Описание датасета: IBM HR Analytics Employee Attrition & Performance

## Общее описание

- Источник: https://www.kaggle.com/datasets/pavansubhasht/ibm-hr-analytics-attrition-dataset/data 
- Размер: (1470,35)
- Признаков: 35 (26 вещественных признаков, 9 - категориальных)
- Целевая переменная: Attrition (строковый тип: Yes - уволился сотрудник, No - не уволился)
- Задача: Классификация, прогнозирование увольнения сотрудника
- Типы данных: Целые числа, булев тип, категории

---

## Целевая переменная 

- Attrition: строковый тип, который является по сути булевым. Признак указывает на то, уволился ли сотрудник из компании
    - Yes - Сотрудник уволился 
    - No - Сотрудник не уволился

```python
import pandas as pd
```

```python
def describe_feat(df, num_cols, cat_cols):
    print(f"{'№':>3} {'НАЗВАНИЕ':<25} {'ВЕЩЕСТВ.':<8} {'КАТЕГОР.':<8} {'ЧИСЛО ЗНАЧЕНИЙ':<17} {'ТИП КОЛОНКИ':<12} {'ЗНАЧЕНИЯ':<40}")
    print("-"*120)

    for idx, col in enumerate(df.columns):
        
        if col in num_cols:
            is_numeric = '+'
        else:
            is_numeric = ''

        if col in cat_cols:
            is_categorical = '+'
        else:
            is_categorical = ''
            
        unique_vals = df[col].unique()
        n_unique = len(unique_vals)
        col_type = df[col].dtype  

        # Не более 5 значений
        if n_unique > 5:
            unique_display = list(unique_vals[:5]) + ["..."]
        else:
            unique_display = list(unique_vals)
        print(f"{idx:>3} {col:<25} {is_numeric:<8} {is_categorical:<8} {n_unique:<17} {str(col_type):<12} {str(unique_display):<40}")
```
